# 9.Groupby Opretions

```python
g = df.groupby('city')
```

```python
g.size()
```

```python
g['age'].mean()   #multi 
```

```python
g.agg({'age':['mean','sum']})  
```

```python
g.apply(lambda x: x['age'].max() - x['age'].min())    #  on a groupby --custom function per group
```

Itrating over group with a for loop

```python
for name,group in g:
    print(name)
    print(group)    
```

.get_group( )   - retrive a specific grourp

```python
g.get_group('NY')
```
